```python
%%writefile app.py
import os
import streamlit as st
import cohere

# 1. Directly assign the API key to bypass Colab environment variable issues
api_key = "BYgDZHL3UItpd4Levo3z4WgfoNxdD1nKzNGOn4jh"

st.set_page_config(
    page_title="College AI Assistant",
    page_icon="🎓"
)

st.title("🎓 College AI Assistant")
st.caption("Your AI-powered academic learning assistant")

if not api_key:
    st.error("Cohere API key is missing. Please ensure it is set.")
    st.stop()
client = cohere.ClientV2(api_key=api_key)

SYSTEM_PROMPT = """
You are a helpful college AI assistant.
Explain academic concepts in simple English.
Give examples and step-by-step explanations.
Answer follow-up questions using conversation history.
If you are unsure, say so honestly.
"""

if "messages" not in st.session_state:
    st.session_state.messages = []

for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

question = st.chat_input("Ask your academic question...")

if question:
    st.session_state.messages.append({"role": "user", "content": question})
    with st.chat_message("user"):
        st.markdown(question)

    with st.chat_message("assistant"):
        try:
            with st.spinner("Thinking..."):
                response = client.chat(
                    model="command-a-plus-05-2026",
                    messages=[
                        {
                            "role": "system",
                            "content": SYSTEM_PROMPT
                        },
                        *st.session_state.messages
                    ]
                )
                answer = "".join(
                    item.text
                    for item in response.message.content
                    if item.type == "text"
                )
            st.markdown(answer)
            st.session_state.messages.append(
                {"role": "assistant", "content": answer}
            )
        except Exception as error:
            st.error(
                "Unable to get an answer. Check your API key, "
                "model access, and internet connection."
            )
            st.caption(str(error))

if st.sidebar.button("Clear conversation"):
    st.session_state.messages = []
    st.rerun()
```
